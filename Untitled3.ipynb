```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Convert pickup time to datetime
df["pickup_datetime"] = pd.to_datetime(
    df["pickup_datetime"],
    errors="coerce"
)

# Remove missing values
df = df.dropna(subset=[
    "pickup_datetime",
    "pickup_latitude",
    "pickup_longitude"
]).copy()

# Remove duplicate trips
df = df.drop_duplicates(subset=["id"])

# Restrict coordinates to the NYC study area
df = df[
    df["pickup_latitude"].between(40.5, 41.0) &
    df["pickup_longitude"].between(-74.3, -73.6)
].copy()

# Extract time features
df["hour"] = df["pickup_datetime"].dt.hour
df["date"] = df["pickup_datetime"].dt.date
df["day_of_week"] = df["pickup_datetime"].dt.day_name()

print("Cleaned dataset shape:", df.shape)
print("Date range:",
      df["pickup_datetime"].min(),
      "to",
      df["pickup_datetime"].max())

display(df.head())
```

```python

hourly_demand = (
    df.groupby("hour")
    .size()
    .reindex(range(24), fill_value=0)
)

plt.figure(figsize=(12, 5))

plt.bar(
    hourly_demand.index,
    hourly_demand.values,
    color="steelblue"
)

plt.xlabel("Hour of Day")
plt.ylabel("Number of Taxi Pickups")
plt.title("Hourly Taxi Pickup Demand in NYC")
plt.xticks(range(24))
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()

print("Highest pickup hour:",
      hourly_demand.idxmax())

print("Number of pickups:",
      hourly_demand.max())
```

```python

plt.figure(figsize=(10, 8))

plt.scatter(
    df["pickup_longitude"],
    df["pickup_latitude"],
    s=2,
    alpha=0.35,
    color="royalblue"
)

plt.xlabel("Pickup Longitude")
plt.ylabel("Pickup Latitude")
plt.title("Spatial Distribution of NYC Taxi Pickups")
plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python

from sklearn.cluster import DBSCAN
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Evening peak: 6 PM to before 9 PM
evening_df = df[df["hour"].between(18, 20)].copy()

# DBSCAN with Haversine distance
coords = np.radians(
    evening_df[["pickup_latitude", "pickup_longitude"]].values
)

earth_radius_km = 6371.0

dbscan = DBSCAN(
    eps=0.5 / earth_radius_km,
    min_samples=10,
    metric="haversine",
    algorithm="ball_tree"
)

evening_df["cluster"] = dbscan.fit_predict(coords)

n_clusters = evening_df.loc[
    evening_df["cluster"] != -1, "cluster"
].nunique()

print("Evening pickup records:", len(evening_df))
print("Number of clusters:", n_clusters)
print("Noise points:", (evening_df["cluster"] == -1).sum())

display(evening_df["cluster"].value_counts().head(10))
```

```python

plt.figure(figsize=(11, 8))

clusters = evening_df["cluster"]

plt.scatter(
    evening_df["pickup_longitude"],
    evening_df["pickup_latitude"],
    c=clusters,
    cmap="tab20",
    s=8,
    alpha=0.7
)

plt.xlabel("Pickup Longitude")
plt.ylabel("Pickup Latitude")
plt.title("DBSCAN Taxi Demand Clusters - Evening Peak")
plt.grid(alpha=0.2)

plt.tight_layout()
plt.savefig("evening_dbscan.png", dpi=300)
plt.show()

```

```python

evening_summary = (
    evening_df[evening_df["cluster"] != -1]
    .groupby("cluster")
    .agg(
        pickup_count=("cluster", "size"),
        center_lat=("pickup_latitude", "mean"),
        center_lon=("pickup_longitude", "mean")
    )
    .sort_values("pickup_count", ascending=False)
)

print("Top evening demand clusters:")
display(evening_summary.head(10))

plt.figure(figsize=(10, 5))

evening_summary.head(10)["pickup_count"].plot(
    kind="bar",
    color="coral"
)

plt.title("Top 10 Evening Taxi Demand Clusters")
plt.xlabel("Cluster ID")
plt.ylabel("Number of Pickups")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("top_evening_clusters.png", dpi=300)
plt.show()
```

```python

from sklearn.cluster import DBSCAN
import numpy as np

# Morning peak: 7 AM to before 10 AM
morning_df = df[df["hour"].between(7, 9)].copy()

morning_coords = np.radians(
    morning_df[["pickup_latitude", "pickup_longitude"]].values
)

morning_dbscan = DBSCAN(
    eps=0.5 / 6371.0,
    min_samples=10,
    metric="haversine",
    algorithm="ball_tree"
)

morning_df["cluster"] = morning_dbscan.fit_predict(
    morning_coords
)

morning_clusters = morning_df.loc[
    morning_df["cluster"] != -1, "cluster"
].nunique()

print("Morning pickup records:", len(morning_df))
print("Morning clusters:", morning_clusters)
print("Noise points:",
      (morning_df["cluster"] == -1).sum())
```

```python

import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Morning
axes[0].scatter(
    morning_df["pickup_longitude"],
    morning_df["pickup_latitude"],
    c=morning_df["cluster"],
    cmap="tab20",
    s=8,
    alpha=0.7
)

axes[0].set_title("Morning Demand Clusters (7 AM - 10 AM)")
axes[0].set_xlabel("Longitude")
axes[0].set_ylabel("Latitude")

# Evening
axes[1].scatter(
    evening_df["pickup_longitude"],
    evening_df["pickup_latitude"],
    c=evening_df["cluster"],
    cmap="tab20",
    s=8,
    alpha=0.7
)

axes[1].set_title("Evening Demand Clusters (6 PM - 9 PM)")
axes[1].set_xlabel("Longitude")
axes[1].set_ylabel("Latitude")

# Same geographic scale
for ax in axes:
    ax.set_xlim(-74.1, -73.75)
    ax.set_ylim(40.6, 40.9)
    ax.grid(alpha=0.2)

plt.suptitle(
    "Dynamic Taxi Demand Zone Comparison",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.savefig(
    "morning_evening_comparison.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python

import pandas as pd

comparison = pd.DataFrame({
    "Time Period": ["Morning", "Evening"],
    "Total Pickups": [
        len(morning_df),
        len(evening_df)
    ],
    "Clusters Detected": [
        morning_clusters,
        n_clusters
    ],
    "Noise Points": [
        (morning_df["cluster"] == -1).sum(),
        (evening_df["cluster"] == -1).sum()
    ]
})

comparison["Clustered Pickups"] = (
    comparison["Total Pickups"] -
    comparison["Noise Points"]
)

comparison["Clustered Percentage"] = (
    100 * comparison["Clustered Pickups"] /
    comparison["Total Pickups"]
).round(2)

display(comparison)
```

```python

comparison.plot(
    x="Time Period",
    y=["Clustered Pickups", "Noise Points"],
    kind="bar",
    stacked=True,
    figsize=(9, 5),
    color=["mediumseagreen", "lightcoral"]
)

plt.title("Morning vs Evening Taxi Pickup Clustering")
plt.ylabel("Number of Pickup Records")
plt.xlabel("Time Period")
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()

plt.savefig(
    "demand_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

from shapely.geometry import MultiPoint
from shapely.ops import transform
from pyproj import Transformer

# NYC local projected coordinates (metres)
project = Transformer.from_crs(
    "EPSG:4326", "EPSG:32618",
    always_xy=True
).transform

def extract_zones(data, period):
    zones = []

    for cluster_id, group in data.groupby("cluster"):
        if cluster_id == -1 or len(group) < 3:
            continue

        points = list(zip(
            group["pickup_longitude"],
            group["pickup_latitude"]
        ))

        hull = MultiPoint(points).convex_hull

        if hull.geom_type != "Polygon":
            continue

        area_km2 = transform(project, hull).area / 1e6

        zones.append({
            "period": period,
            "cluster": cluster_id,
            "pickups": len(group),
            "area_km2": area_km2,
            "density": len(group) / max(area_km2, 0.01),
            "center_lat": group["pickup_latitude"].mean(),
            "center_lon": group["pickup_longitude"].mean(),
            "geometry": hull
        })

    return zones

morning_zones = extract_zones(morning_df, "Morning")
evening_zones = extract_zones(evening_df, "Evening")

print("Morning zone boundaries:", len(morning_zones))
print("Evening zone boundaries:", len(evening_zones))
```

```python

import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 9))

# Plot evening pickup points
ax.scatter(
    evening_df["pickup_longitude"],
    evening_df["pickup_latitude"],
    s=3,
    color="royalblue",
    alpha=0.3,
    label="Taxi Pickup Locations"
)

# Draw cluster boundaries
for zone in evening_zones:
    polygon = zone["geometry"]

    x, y = polygon.exterior.xy

    ax.fill(
        x, y,
        color="red",
        alpha=0.15
    )

    ax.plot(
        x, y,
        color="darkred",
        linewidth=1.5
    )

ax.set_title(
    "DBSCAN Evening Taxi Demand Zones",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlabel("Pickup Longitude")
ax.set_ylabel("Pickup Latitude")
ax.set_aspect("equal", adjustable="box")
ax.grid(alpha=0.2)
ax.legend()

plt.tight_layout()

plt.savefig(
    "evening_demand_zones.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Demand zone map saved successfully!")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

zone_table = pd.DataFrame([
    {
        "Cluster": z["cluster"],
        "Pickups": z["pickups"],
        "Area_km2": round(z["area_km2"], 3),
        "Density": round(z["density"], 2)
    }
    for z in evening_zones
])

zone_table = zone_table.sort_values(
    "Pickups", ascending=False
).reset_index(drop=True)

# Dispatch priority based on historical pickup counts
zone_table["Dispatch_Priority"] = [
    "HIGH" if i < 3 else
    "MEDIUM" if i < 7 else
    "LOW"
    for i in range(len(zone_table))
]

print("Top taxi demand zones:")
display(zone_table.head(10))
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Rebuild the zone table from detected clusters
zone_table = pd.DataFrame([
    {
        "Cluster": z["cluster"],
        "Pickups": z["pickups"],
        "Area_km2": round(z["area_km2"], 3),
        "Density": round(z["density"], 2)
    }
    for z in evening_zones
])

if zone_table.empty:
    raise ValueError(
        "No evening cluster zones found. "
        "Check your DBSCAN and boundary extraction steps."
    )

zone_table = zone_table.sort_values(
    "Pickups", ascending=False
).reset_index(drop=True)

# Simulated available drivers
rng = np.random.default_rng(42)

zone_table["Simulated_Drivers"] = rng.integers(
    10, 51, size=len(zone_table)
)

# Create the missing column
zone_table["Demand_Supply_Ratio"] = (
    zone_table["Pickups"] /
    zone_table["Simulated_Drivers"]
).round(2)

# Simulated surge categories
zone_table["Surge_Level"] = np.select(
    [
        zone_table["Demand_Supply_Ratio"] >= 10,
        zone_table["Demand_Supply_Ratio"] >= 5
    ],
    ["HIGH", "MEDIUM"],
    default="LOW"
)

# Display results
print("Surge Simulation Results:")
display(zone_table.head(10))

# Plot surge simulation
surge_colors = {
    "HIGH": "red",
    "MEDIUM": "orange",
    "LOW": "green"
}

top_surge = zone_table.head(10)

plt.figure(figsize=(11, 6))

plt.bar(
    top_surge["Cluster"].astype(str),
    top_surge["Demand_Supply_Ratio"],
    color=[
        surge_colors[level]
        for level in top_surge["Surge_Level"]
    ]
)

plt.title("Illustrative Taxi Surge Demand Indicator")
plt.xlabel("DBSCAN Cluster ID")
plt.ylabel("Historical Pickups / Simulated Drivers")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("surge_simulation.png", dpi=300)
plt.show()

# Export results
zone_table.to_csv(
    "final_taxi_demand_results.csv",
    index=False
)

print("SUCCESS! Simulation graph and CSV saved.")

```

```python

import pandas as pd
import matplotlib.pyplot as plt

def evaluate_dbscan(data, period):
    total = len(data)
    noise = (data["cluster"] == -1).sum()
    clusters = data.loc[
        data["cluster"] != -1, "cluster"
    ].nunique()

    return {
        "Period": period,
        "Total Pickups": total,
        "Clusters": clusters,
        "Noise Points": noise,
        "Clustered (%)": round(
            100 * (total - noise) / total, 2
        ) if total else 0
    }

evaluation = pd.DataFrame([
    evaluate_dbscan(morning_df, "Morning"),
    evaluate_dbscan(evening_df, "Evening")
])

print("FINAL DBSCAN EVALUATION")
display(evaluation)

evaluation.to_csv(
    "dbscan_evaluation.csv",
    index=False
)
```

```python

fig, ax = plt.subplots(figsize=(9, 5))

bars = ax.bar(
    evaluation["Period"],
    evaluation["Clustered (%)"],
    color=["royalblue", "tomato"],
    width=0.5
)

ax.bar_label(bars, fmt="%.1f%%", padding=3)

ax.set_ylim(0, 110)
ax.set_ylabel("Clustered Pickup Percentage")
ax.set_title("DBSCAN Clustering Coverage: Morning vs Evening")
ax.grid(axis="y", alpha=0.2)

plt.tight_layout()
plt.savefig(
    "dbscan_evaluation.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
